```python
import polars as pl
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pl.read_csv('./scores_transformer.tsv', separator='\t')
data = data.with_columns(pl.col('CHECK').str.split('_').list.last().cast(pl.Int32))

plt.figure(figsize=(8, 6))

plt.hlines(y=.55, xmin=0, xmax=100, colors='black', label='Full Dataset - Scratch')
sns.set_theme('talk')
sns.lineplot(
    data=data,
    x='CHECK',
    y='CHAMMI',
    style='MODEL',
    hue='MODEL',
    palette=['#610F7F', '#7A8450'],
    linewidth=3
)
plt.title("Transformer Score Versus Checkpoint")
```

```python
loss_csv = "./transformer_loss_curve.csv"
loss_csv = pl.read_csv(loss_csv)
loss_csv = loss_csv['Step', 'scratch', 'finetuned']
loss_csv = loss_csv.unpivot(index=["Step"], variable_name='Model', value_name='Loss')
sns.set_theme('talk')
plt.figure(figsize=(12, 12))

sns.lineplot(
    data=loss_csv,
    x='Step',
    y='Loss',
    style='Model',
    hue='Model',
    palette=['#610F7F', '#7A8450'],
    linewidth=3
)

size = 30
plt.xticks(fontsize=size)
plt.yticks(fontsize=size)
plt.legend(fontsize=size)
plt.xlabel("Iteration", fontsize=size)
plt.ylabel("Loss", fontsize=size)
plt.title("Transformer Score Versus Checkpoint", fontsize=size)
```
