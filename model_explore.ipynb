```python
import torch
from transformers import AutoImageProcessor, AutoModel
```

```python
mod_path = '/hdd/jcaicedo/projects/dinov3/ddw2026/models/scratch_100epoch/eval/checkpoint_100.pt'
state_dict = torch.load(mod_path, map_location="cpu")

state_dict['model'].keys()
```

```python
backbone = AutoModel.from_pretrained("facebook/dinov3-vits16-pretrain-lvd1689m", device_map="auto")
```

```python
new_state_dict = {}
for k, v in state_dict['model'].items():
    if 'predictor' in k: continue
    name = k.replace("backbone.", "") if k.startswith("backbone.") else k
    new_state_dict[name] = v
backbone.load_state_dict(new_state_dict, strict=True)
```
